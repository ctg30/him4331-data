# 🪜 M04.4 — Data Quality Check *(10 pts)*
**This week you:** find one real data-quality issue, fix it by choosing from a menu (no code editing), and defend your choice.

**💡 Known trouble spots in YOUR file** — I've checked all four, so these are real:

- **diabetes** → four columns hide missing values as `"?"`: `weight` (97% missing!), `medical_specialty` (~49%), `payer_code` (~40%), `race` (~2%). Also `max_glu_serum` and `A1Cresult` are genuinely blank for most patients. → try fix `recode_missing`, or `drop_rows` if it's a column you must have.
- **hospitals** → `Hospital overall rating` is stored as TEXT because ~1,000 hospitals say `Not Available` instead of a 1–5 star number. You cannot do math on it until it's converted. → fix `to_number`.
- **readmissions** → `Number of Discharges` is blank on 10,088 of 18,330 rows, and `Excess Readmission Ratio` is blank on 6,610 (CMS suppresses small-volume hospitals). → fix `drop_rows` on the column your question needs.
- **breaches** → there are exactly **2 duplicate rows** (→ fix `drop_duplicates`), and `Individuals Affected` has wild outliers: the median breach affects ~6,500 people but the largest affects 192.7 million, which will distort every average you compute. → `document_only` is a defensible choice here if you explain the outlier.

**You're done when:** the detective cell shows a problem, your fix menu passes its self-check, the AFTER cell proves the fix, and the 📤 M04.4 Submission Block prints cleanly.

```python
#@title ▶ STEP 1: The Detective Cell — finds candidate problems for you (just run it)
try:
    print("🔍 A) MISSING VALUES per column (top 15):")
    print(df.isna().sum().sort_values(ascending=False).head(15).to_string())
    print()
    print("🔍 B) DUPLICATE ROWS:", df.duplicated().sum())
    print()
    print("🔍 C) COLUMNS THAT MIGHT HIDE MISSING VALUES AS CODES like '?', 'Unknown', 'Not Available', 'N/A':")
    suspects = []
    for col in df.columns:
        if not pd.api.types.is_numeric_dtype(df[col]):
            vals = df[col].astype(str)
            hits = vals.isin(["?", "Unknown", "UNKNOWN", "Not Available", "N/A", "NA", "None", ""]).sum()
            if hits > 0:
                suspects.append((col, int(hits)))
    for col, hits in sorted(suspects, key=lambda x: -x[1])[:10]:
        print(f"   {col}: {hits:,} suspicious values")
    if not suspects:
        print("   (none found — focus on A or B above)")

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

```python
#@title 🟨 TODO — STEP 2: name the column and pick your fix from the menu
try:
    M04_COLUMN = ""   # the column with the issue, copied exactly from the detective output
    M04_FIX    = ""   # pick ONE (type it exactly):
                      #   "recode_missing"  → turn codes like '?' / 'Not Available' into proper missing values
                      #   "drop_rows"       → remove rows that are missing this column
                      #   "impute_median"   → fill missing numbers with the column's median
                      #   "to_number"       → convert a text column to numbers (bad values become missing)
                      #   "drop_duplicates" → remove duplicate rows (M04_COLUMN still required, for reporting)
                      #   "document_only"   → keep the data as-is and explain why that's the right call

    _FIX_MENU = ["recode_missing", "drop_rows", "impute_median", "to_number",
                 "drop_duplicates", "document_only"]
    check(M04_COLUMN in df.columns, "Column found.",
          "M04_COLUMN must be an exact column name from the detective output above.")
    check(M04_FIX in _FIX_MENU, f"Fix choice '{M04_FIX}' is valid.",
          "M04_FIX must be exactly one of: " + ", ".join(_FIX_MENU))

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

```python
#@title ▶ STEP 3: BEFORE → apply your fix → AFTER (just run it — the notebook does the coding)
try:
    _MISS_CODES = ["?", "Unknown", "UNKNOWN", "Not Available", "N/A", "NA", "None", ""]

    def _diagnose(frame, col):
        s = frame[col]
        coded = s.astype(str).isin(_MISS_CODES).sum() if not pd.api.types.is_numeric_dtype(s) else 0
        return (f"missing={s.isna().sum():,} | coded-as-missing={coded:,} | "
                f"duplicated rows in table={frame.duplicated().sum():,} | dtype={s.dtype}")

    print("BEFORE:", _diagnose(df, M04_COLUMN))

    df_clean = df.copy()
    if M04_FIX == "recode_missing":
        df_clean[M04_COLUMN] = df_clean[M04_COLUMN].replace(_MISS_CODES, np.nan)
    elif M04_FIX == "drop_rows":
        df_clean[M04_COLUMN] = df_clean[M04_COLUMN].replace(_MISS_CODES, np.nan)
        df_clean = df_clean.dropna(subset=[M04_COLUMN])
    elif M04_FIX == "impute_median":
        df_clean[M04_COLUMN] = pd.to_numeric(df_clean[M04_COLUMN], errors="coerce")
        df_clean[M04_COLUMN] = df_clean[M04_COLUMN].fillna(df_clean[M04_COLUMN].median())
    elif M04_FIX == "to_number":
        df_clean[M04_COLUMN] = pd.to_numeric(
            df_clean[M04_COLUMN].replace(_MISS_CODES, np.nan), errors="coerce")
    elif M04_FIX == "drop_duplicates":
        df_clean = df_clean.drop_duplicates()
    elif M04_FIX == "document_only":
        pass  # keeping data as-is, on purpose

    print("AFTER: ", _diagnose(df_clean, M04_COLUMN))
    print(f"Rows: {len(df):,} → {len(df_clean):,}")
    print()
    print("✅ From here on, every later section automatically uses your cleaned data (df_clean).")

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

### 🔧 Optional helper — turn a number into a yes/no group

**Skip this unless you need it.** Some questions need a *category* where your data only has a *number*. Example: "which hospitals perform WORSE than expected?" — that's a yes/no group built from the number `Excess Readmission Ratio` (above 1.0 = worse).

You'll want this if:
- your dataset is **readmissions** (needed — otherwise M05 has no meaningful category pair), or
- your dataset is **hospitals** or **breaches** and you want a yes/no outcome, or
- you plan to run a **logistic** regression in M07 (it needs a yes/no outcome).

**Skip it if** your dataset is **diabetes** — you already have `readmitted`, `change`, `diabetesMed`, and `insulin` as ready-made categories.

Leave all four boxes empty to skip. Nothing breaks either way.

**Ready-made settings you can copy:**
- readmissions → column `Excess Readmission Ratio`, cutoff `1.0`, yes-label `Worse than expected`, no-label `As expected or better`
- breaches → column `Individuals Affected`, cutoff `10000`, yes-label `Large breach`, no-label `Smaller breach`

```python
#@title 🟨 OPTIONAL — make a yes/no column called MY_FLAG (leave blank to skip)
try:
    FLAG_FROM_COLUMN = ""   # a NUMBER column, e.g. "Excess Readmission Ratio"
    FLAG_ABOVE       = ""   # the cutoff, in quotes, e.g. "1.0"
    FLAG_YES_LABEL   = ""   # label for rows ABOVE the cutoff, e.g. "Worse than expected"
    FLAG_NO_LABEL    = ""   # label for rows at or below,     e.g. "As expected or better"

    if FLAG_FROM_COLUMN == "" and FLAG_ABOVE == "":
        print("⏭️  Skipped — you don't need MY_FLAG. Carry on to M05.")
    elif FLAG_FROM_COLUMN not in df_clean.columns:
        print(f"🟨 '{FLAG_FROM_COLUMN}' isn't a column in your dataset. Copy the exact name, or clear all four boxes to skip.")
    else:
        _num = pd.to_numeric(df_clean[FLAG_FROM_COLUMN], errors="coerce")
        try:
            _cut = float(FLAG_ABOVE)
        except ValueError:
            _cut = None
            print(f"🟨 FLAG_ABOVE ('{FLAG_ABOVE}') needs to be a number in quotes, like \"1.0\".")
        if _cut is not None:
            _yes = FLAG_YES_LABEL or "Above cutoff"
            _no = FLAG_NO_LABEL or "At or below cutoff"
            _flag = pd.Series([None] * len(df_clean), index=df_clean.index, dtype=object)
            _flag[_num.notna() & (_num > _cut)] = _yes
            _flag[_num.notna() & (_num <= _cut)] = _no
            df_clean["MY_FLAG"] = _flag
            print(f"✅ Created a new column called MY_FLAG from '{FLAG_FROM_COLUMN}' (cutoff {_cut}).")
            print("   Use \"MY_FLAG\" as a variable name in M05, M06, or M07. Counts:")
            print(df_clean["MY_FLAG"].value_counts(dropna=False).to_string())

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

```python
#@title ▶ Group summary + two-sample t-test (just run it)
try:
    summary_m06 = df_clean.groupby(GROUP_VAR)[NUM_VAR].agg(["count", "mean", "std"]).round(2)
    print("GROUP SUMMARY:")
    print(summary_m06.to_string())
    print()
    # the test compares the two LARGEST groups (works even if GROUP_VAR has 3+ categories)
    top2 = df_clean[GROUP_VAR].value_counts().index[:2]
    g1 = df_clean.loc[df_clean[GROUP_VAR] == top2[0], NUM_VAR].dropna()
    g2 = df_clean.loc[df_clean[GROUP_VAR] == top2[1], NUM_VAR].dropna()
    t_stat, p_value_m06 = stats.ttest_ind(g1, g2, equal_var=False)
    print(f"Comparing '{top2[0]}' (n={len(g1):,}) vs '{top2[1]}' (n={len(g2):,})")
    print(f"t = {t_stat:.3f}   p-value = {p_value_m06:.4f}")
    print(f"Difference in means = {g1.mean() - g2.mean():.3f}")
    print()
    print("📖 How to read this: p BELOW 0.05 = the difference between the two group")
    print("   averages is probably real, not luck. Also look at the SIZE of the")
    print("   difference — a 'real' difference can still be too small to matter.")

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

# 🪜 M07.4 — Regression *(20 pts)*
**This week you:** fit one regression that speaks directly to YOUR research question, then interpret ONE coefficient in plain English. The notebook writes all the code — you fill in a menu.

**Which model type?** Simple rule: if your outcome is a **number** (length of stay, individuals affected, a ratio) → `"linear"`. If your outcome is a **yes/no** (readmitted or not) → `"logistic"`.

**💡 Models that work in YOUR file** — copy one of these exactly:

- **diabetes, logistic** → MODEL_TYPE `logistic`, OUTCOME `readmitted`, OUTCOME_YES `<30`, PREDICTORS `time_in_hospital` + `num_medications`
- **hospitals, linear** → OUTCOME `Hospital overall rating` *(needs the M04 `to_number` fix first)*, PREDICTORS `Hospital Ownership` + `Emergency Services`
- **readmissions, linear** → OUTCOME `Excess Readmission Ratio`, PREDICTORS `Number of Discharges` + `Measure Name`
- **readmissions, logistic** → OUTCOME `MY_FLAG`, OUTCOME_YES `Worse than expected`, PREDICTORS `Number of Discharges` + `Measure Name`
- **breaches, linear** → OUTCOME `Individuals Affected`, PREDICTORS `Type of Breach` + `Business Associate Present`

Column names with spaces are fine — the notebook handles them for you.

```python
#@title 🟨 TODO — fill in the model menu (no code to write)
try:
    MODEL_TYPE  = ""         # "linear" (number outcome)  or  "logistic" (yes/no outcome)
    OUTCOME     = ""         # the column you're predicting, e.g. "time_in_hospital"
    PREDICTORS  = ["", ""]   # 1–3 predictor columns, each in quotes (delete a "" if you only want 2)
    OUTCOME_YES = ""         # ONLY for logistic: which value of OUTCOME counts as "yes"? e.g. "<30"
                             # (leave "" for linear)

    check(MODEL_TYPE in ("linear", "logistic"), f"Model type '{MODEL_TYPE}' is valid.",
          'MODEL_TYPE must be exactly "linear" or "logistic".')
    _all = [OUTCOME] + [p for p in PREDICTORS if p]
    _missing = [v for v in _all if v not in df_clean.columns]
    check(OUTCOME != "" and len(_missing) == 0, "All variables found.",
          f"Not valid column names: {_missing or '(fill in OUTCOME and PREDICTORS)'}")
    if MODEL_TYPE == "logistic" and OUTCOME in df_clean.columns:
        check(OUTCOME_YES in df_clean[OUTCOME].astype(str).unique(),
              f"'{OUTCOME_YES}' found in {OUTCOME}.",
              f"OUTCOME_YES must be one of: {list(df_clean[OUTCOME].astype(str).unique()[:8])}")
    if MODEL_TYPE == "linear" and OUTCOME in df_clean.columns:
        check(pd.api.types.is_numeric_dtype(df_clean[OUTCOME]),
              f"'{OUTCOME}' is numeric — good for linear.",
              f"'{OUTCOME}' isn't a number column. Either pick a numeric outcome, use M04's "
              "'to_number' fix on it, or switch MODEL_TYPE to \"logistic\".")

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

```python
#@title ▶ Fit the model (just run it — the notebook writes the code)
try:
    import statsmodels.formula.api as smf

    _dfm = df_clean.copy()
    _terms = []
    for _p in [p for p in PREDICTORS if p]:
        _terms.append(f"Q('{_p}')" if pd.api.types.is_numeric_dtype(_dfm[_p]) else f"C(Q('{_p}'))")

    if MODEL_TYPE == "linear":
        formula = f"Q('{OUTCOME}') ~ " + " + ".join(_terms)
        model = smf.ols(formula, data=_dfm).fit()
        fit_label, fit_value = "R-squared", model.rsquared
    else:
        _dfm["_outcome01"] = (_dfm[OUTCOME].astype(str) == OUTCOME_YES).astype(int)
        formula = "_outcome01 ~ " + " + ".join(_terms)
        model = smf.logit(formula, data=_dfm).fit(disp=0)
        fit_label, fit_value = "Pseudo R-squared", model.prsquared

    print("MODEL:", MODEL_TYPE, "regression —", formula)
    print(model.summary())
    print()
    print("📖 How to read this: find your predictor's row in the middle table.")
    print("   'coef' = how much the outcome changes per 1-unit increase in that predictor.")
    print("   'P>|t|' (or 'P>|z|') = that coefficient's p-value — below 0.05 means the")
    print("   effect is statistically convincing.")

except (KeyError, NameError, AttributeError, TypeError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```
