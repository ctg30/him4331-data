## 🟨 Step 4: this week's work
**Which model type?** Simple rule: if your outcome is a **number** (length of stay, individuals affected, a ratio) → `"linear"`. If your outcome is a **yes/no** (readmitted or not) → `"logistic"`.

💡 **Models that work in YOUR file.** Copy one of these exactly:

* **diabetes, linear** → MODEL_TYPE `linear`, OUTCOME `time_in_hospital`, PREDICTORS `num_medications` + `age`
* **diabetes, logistic** → MODEL_TYPE `logistic`, OUTCOME `readmitted`, OUTCOME_YES `<30`, PREDICTORS `time_in_hospital` + `num_medications`
* **hospitals, linear** → OUTCOME `Hospital overall rating` (needs `to_number` as your M04 fix), PREDICTORS `Hospital Ownership` + `Emergency Services`
* **readmissions, linear** → OUTCOME `Excess Readmission Ratio`, PREDICTORS `Number of Discharges` + `Measure Name`
* **readmissions, logistic** → OUTCOME `MY_FLAG`, OUTCOME_YES `Worse than expected`, PREDICTORS `Number of Discharges` + `Measure Name`
* **breaches, linear** → OUTCOME `Individuals Affected`, PREDICTORS `Type of Breach` + `Business Associate Present`

Column names with spaces are fine. The notebook handles them for you. Rows with a blank outcome are left out of the model automatically.

```python
#@title 🟨 TODO: fill in the model menu (no code to write)
try:
    MODEL_TYPE  = ""         # "linear" (number outcome)  or  "logistic" (yes/no outcome)
    OUTCOME     = ""         # the column you're predicting, e.g. "time_in_hospital"
    PREDICTORS  = ["", ""]   # 1 to 3 predictor columns, each in quotes (delete a "" if you only want 2)
    OUTCOME_YES = ""         # ONLY for logistic: which value of OUTCOME counts as "yes"? e.g. "<30"
                             # (leave "" for linear)

    MODEL_TYPE, OUTCOME, OUTCOME_YES = MODEL_TYPE.strip().lower(), OUTCOME.strip(), OUTCOME_YES.strip()
    PREDICTORS = [p.strip() for p in PREDICTORS if p.strip()]
    check(MODEL_TYPE in ("linear", "logistic"), f"Model type '{MODEL_TYPE}' is valid.",
          'MODEL_TYPE must be exactly "linear" or "logistic".')
    _all = [OUTCOME] + PREDICTORS
    _missing = [v for v in _all if v not in df_clean.columns]
    check(OUTCOME != "" and len(PREDICTORS) >= 1 and len(_missing) == 0, "All variables found.",
          f"Not valid column names: {_missing or '(fill in OUTCOME and at least one PREDICTOR)'}")
    check(OUTCOME not in PREDICTORS, "Outcome is not also a predictor.",
          "OUTCOME cannot also appear in PREDICTORS.")
    if MODEL_TYPE == "logistic" and OUTCOME in df_clean.columns:
        check(OUTCOME_YES in df_clean[OUTCOME].dropna().astype(str).unique(),
              f"'{OUTCOME_YES}' found in {OUTCOME}.",
              f"OUTCOME_YES must be one of: {list(df_clean[OUTCOME].dropna().astype(str).unique()[:8])}")
    if MODEL_TYPE == "linear" and OUTCOME in df_clean.columns:
        check(pd.api.types.is_numeric_dtype(df_clean[OUTCOME]),
              f"'{OUTCOME}' is numeric, good for linear.",
              f"'{OUTCOME}' isn't a number column. Either pick a numeric outcome, set M04_FIX to "
              "'to_number' on it in Step 2, or switch MODEL_TYPE to \"logistic\".")
    for _p in PREDICTORS:
        if _p in df_clean.columns and not pd.api.types.is_numeric_dtype(df_clean[_p]):
            check(df_clean[_p].nunique() <= 60,
                  f"'{_p}' is a category with {df_clean[_p].nunique()} groups (fine).",
                  f"'{_p}' has {df_clean[_p].nunique()} different values, too many for a readable model. Pick another predictor.")

except (KeyError, NameError, AttributeError, TypeError, IndexError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
```

```python
#@title ▶ Fit the model (just run it; the notebook writes the code)
try:
    import statsmodels.formula.api as smf

    _dfm = df_clean.dropna(subset=[OUTCOME] + PREDICTORS).copy()
    _terms = []
    for _p in PREDICTORS:
        _terms.append(f"Q('{_p}')" if pd.api.types.is_numeric_dtype(_dfm[_p]) else f"C(Q('{_p}'))")

    if MODEL_TYPE == "linear":
        formula = f"Q('{OUTCOME}') ~ " + " + ".join(_terms)
        model = smf.ols(formula, data=_dfm).fit()
        fit_label, fit_value = "R-squared", model.rsquared
    else:
        _dfm["_outcome01"] = (_dfm[OUTCOME].astype(str) == OUTCOME_YES).astype(int)
        formula = "_outcome01 ~ " + " + ".join(_terms)
        model = smf.logit(formula, data=_dfm).fit(disp=0, maxiter=200)
        fit_label, fit_value = "Pseudo R-squared", model.prsquared

    print("MODEL:", MODEL_TYPE, "regression:", formula)
    print(f"Rows used: {len(_dfm):,} of {len(df_clean):,} (rows with a blank outcome or predictor are left out)")
    print(model.summary())
    print()
    print("📖 How to read this: find your predictor's row in the middle table.")
    print("   'coef' = how much the outcome changes per 1-unit increase in that predictor.")
    if MODEL_TYPE == "logistic":
        print("   (Logistic: a positive coef means the 'yes' outcome gets MORE likely as the predictor rises.)")
    print("   'P>|t|' (or 'P>|z|') = that coefficient's p-value. Below 0.05 means the")
    print("   effect is statistically convincing.")

except (KeyError, NameError, AttributeError, TypeError, IndexError):
    print("🟨 This cell needs an earlier cell finished first.")
    print("   Scroll up, complete the 🟨 TODO cells above in order, then re-run this cell.")
    print("   (Fastest fix: Runtime → Run all)")
except Exception as e:
    print("🟨 Something unexpected happened in this cell. Try Runtime → Run all first.")
    print("   If it still fails, screenshot this and message your instructor:", repr(e))
```
